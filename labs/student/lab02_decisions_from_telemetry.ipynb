## 6. Boundary cases

A boundary is a value at which the required outcome changes. Boundary mistakes often come from choosing `<` when the requirement means `<=`, or the reverse.

Requirement:

> A warning is required when wheel speed is at least 6000 rpm.

### WRITE IT YOURSELF 2

Complete the comparison, then test 5999, 6000, and 6001 rpm. Explain why all three tests are useful.

````python
wheel_speed_rpm = 6000
for wheel_speed_rpm in [5999, 6000, 6001]:
    warning_required = wheel_speed_rpm >= 6000
    print(f"{wheel_speed_rpm}: {warning_required}")
````

**Why test all three values?**

Testing all three values verifies the exact boundary condition: 5999 checks just below the threshold (should be False), 6000 checks the exact boundary value itself (should be True), and 6001 checks just above the threshold (should be True). This ensures the correct comparison operator (`>=`) was used rather than a strict inequality.

## 8. Implement the telemetry classifier

Use the four input variables below and an `if`/`elif`/`else` chain. Assign exactly one of these strings to `status`:

```text
CRITICAL
SAFE MODE
WARNING
NOMINAL
```

Do not start coding until your decision table is complete.

````python
battery_charge = 80
bus_voltage_v = 28
panel_temperature_c = 20
comms_ok = True

# Implement the priority rules here.
if battery_charge < 15 or bus_voltage_v < 22:
    status = "CRITICAL"
elif battery_charge < 25 or bus_voltage_v < 24 or panel_temperature_c < -20 or panel_temperature_c > 85 or not comms_ok:
    status = "SAFE MODE"
elif battery_charge <= 35 or panel_temperature_c > 75:
    status = "WARNING"
else:
    status = "NOMINAL"

print("Spacecraft status:", status)
````

### Explain your implementation

Why must the CRITICAL test appear before the SAFE MODE test? Why is `else` appropriate for NOMINAL?

**Your explanation:**

The CRITICAL test must appear before SAFE MODE because any condition that triggers CRITICAL (such as battery charge below 15%) also satisfies the wider threshold for SAFE MODE (below 25%). If SAFE MODE were evaluated first, a critical failure would incorrectly trigger SAFE MODE instead of the highest-priority CRITICAL status, violating the rule that the first applicable state wins. 

`else` is appropriate for NOMINAL because NOMINAL represents the default fallback state when none of the specific higher-priority hazard conditions (CRITICAL, SAFE MODE, or WARNING) apply.

#### 1. Chosen Inputs & 2. Predicted Outcomes:

* **Test 4 (At 24 V):** `battery = 80, voltage = 24, temp = 20, comms = True` $\rightarrow$ **Expected:** `NOMINAL` (Voltage is no longer < 24 V, and all other parameters are normal).
* **Test 5 (Above 24 V):** `battery = 80, voltage = 25, temp = 20, comms = True` $\rightarrow$ **Expected:** `NOMINAL` (Well above all failure thresholds).

#### 4. Explanation:
These tests prove that the strict inequality operators (`< 22` and `< 24`) correctly separate the operational states at the exact boundary points. Specifically, a voltage of 21 V drops the spacecraft to `CRITICAL`, 22 V and 23 V correctly step up to `SAFE MODE`, and 24 V successfully clears the fault condition to return the system to `NOMINAL`.
